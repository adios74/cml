7. Рассмотрим случайный вектор с формой 100х2, представляющей координаты, и найдем расстояния от точки к точке

````python
arr = np.random.rand(100, 2)
arr = np.array(arr)

distances = []
for i in range(1, 100):
    point1 = arr[i - 1]
    point2 = arr[i]
    square = np.square(point2 - point1)
    distance = np.sqrt(np.sum(square))
    distances.append(distance)
print(distances)
````

````python
import pandas as pd
````

````python
DATA_PATH = 'russian-housing-market.csv'
df = pd.read_csv(DATA_PATH)
df.head()
````

1. Посчитайте количество записей в датасете с отсутствующей информацией о жилой площади жилого помещения

````python
total_count = len(df["life_sq"])
not_null_count = df["life_sq"].count()
print(total_count - not_null_count)
````

2. Предложите способ заменить отсутствующие значения из пункта 1 и реализуйте его. Способ нужно аргументировать(сказать почему это достаточно хорошее приближение)

````python
# Подставляем произведение площади всей квартиры на медианное соотношение жилой площади к площади всей квартиры.
# Медиана выбрана, потому что она позволяет добиться типичного соотношения площадей квартир.
df["life_sq"] = df["life_sq"].fillna(df["full_sq"] * (df["life_sq"] / df["full_sq"]).median())
````

9. Проанализируйте есть ли разница в цене между квартирами на 1 и 9 этажах

### 3. Визуализация

Используя библиотеки matplotlib/sns/plotly/pandas и набор данных представленный ранее визуализируйте

1. Как менялась средняя цена квартиры в Нагатинском Затоне(Nagatinski Zaton) с течением времени

````python
import matplotlib.pyplot as plt

data = df[df["sub_area"] == "Nagatinskij Zaton"].copy()
data["timestamp"] = pd.to_datetime(data["timestamp"])
price_month = data.set_index("timestamp").resample("Y")["price_doc"].mean()
price_month /= 1000000

fig, ax = plt.subplots(figsize = (10, 4))
ax.plot(price_month.index, price_month.values, linewidth=2, label="Средняя цена квартир", color="steelblue")
ax.set(title="Средняя цена квартир в Нагатинском Затоне", xlabel = "Дата", ylabel = "Цена, млн руб")
ax.legend()
ax.grid(alpha = 0.2)
fig.autofmt_xdate()
plt.show()
````

2. Постройте гистограмму распределения цен на квартиры с общей площадью более 50 кв.метров

````python
data = df[(df["full_sq"] > 50) & (df["price_doc"].notna())]["price_doc"]
data /= 1000000
fig, ax = plt.subplots(figsize = (10, 5))
ax.hist(data, bins=100, color="steelblue", edgecolor="white")
ax.axvline(data.median(), color="tomato", linestyle="--", label="Медиана")
ax.set(title="Распределение цены", xlabel="Цена, млн руб", ylabel="Кол-во квартир")
ax.set_xticks(np.arange(0, 121, 5))
ax.legend()
plt.show()
````

3. Визуализируйте количество проданных квартир в разных районах

````python
data = df["sub_area"].value_counts()

plt.figure(figsize=(30, 5))
plt.bar(data.index, data.values)
plt.title("Кол-во проданных квартир по районам")
plt.xlabel("Район")
plt.ylabel("Кол-во проданных квартир")
plt.xticks(rotation=90)
plt.show()
````

4. Постройте box plot отображающий распределение полных площадей квартир в Басманном районе

````python
data = df[df["sub_area"] == "Basmannoe"]["full_sq"].dropna()
plt.figure(figsize=(10, 5))
plt.boxplot(data)
plt.title("Распределение полных площадей квартир в Басманном районе")
plt.ylabel("Полные площади квартир")
plt.show()
````

5. Визуализируйте зависимость средней стоимости квартир в зависимости от этажа

````python
data = df.groupby("floor")["price_doc"].mean()
data /= 1000000

fig, ax = plt.subplots(figsize=(10, 5))
ax.scatter(data.index, data.values, alpha=0.5, s=35)
ax.set(title="Зависимость средней цены стоимости квартир от этажа", xlabel="Этаж", ylabel="Средняя цена, млн руб")
ax.grid(alpha=0.2)
plt.show()
````

6. Покажите как меняется средняя площадь квартир в зависимости от района

````python
data = df.groupby("sub_area")["full_sq"].mean().sort_values()

fig, ax = plt.subplots(figsize=(20, 30))
ax.barh(data.index, data.values, color="teal")
ax.set(title="Зависимость средней площади квартиры от района", xlabel="Средняя площадь, кв м", ylabel="Район")
ax.grid(alpha=0.2)
plt.show()
````

7. Постройте график изменения медианной цены на квартиру менее 50 кв. метров за период с 2011 до 2014 года*

\* Задание на доп. балл

````python
data = df[df["full_sq"] < 50].copy()
data["timestamp"] = pd.to_datetime(data["timestamp"])
data = data[(data["timestamp"].dt.year >= 2011) & (data["timestamp"].dt.year <= 2014)]
price_month = data.set_index("timestamp").resample("M")["price_doc"].median()
price_month /= 1000000

fig, ax = plt.subplots(figsize = (10, 4))
ax.plot(price_month.index, price_month.values, linewidth=2, label="Медианная цена квартир", color="steelblue")
ax.set(title="Медианная цена квартиры с площадью менее 50 кв м в период с 2011 до 2014 года", xlabel = "Дата", ylabel = "Цена, млн руб")
ax.legend()
ax.grid(alpha = 0.2)
fig.autofmt_xdate()
plt.show()
````
